### Exercise 1.4 — Boolean indexing

The marks of 12 students are given below. Using boolean indexing (no loops):
1. print the marks that are 40 or more (pass)
2. count how many students failed
3. print the marks between 60 and 80 (both inclusive)

```python
marks = np.array([35, 72, 48, 91, 27, 66, 54, 80, 39, 77, 60, 45])
print("Passed marks:  ", marks[marks >= 40])
print("Number failed: ", (marks < 40).sum())
print("Between 60-80: ", marks[(marks >= 60) & (marks <= 80)])
```

### Exercise 1.5 — Vectorised operations

Convert the week's maximum temperatures from Celsius to Fahrenheit (F = C × 9/5 + 32) **without a loop**. Then find which day was the hottest and by how many degrees Celsius each day differs from the weekly mean.

```python
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
temps_c = np.array([31.2, 33.5, 35.1, 34.8, 29.9, 30.4, 32.7])

temps_f = temps_c * 9 / 5 + 32
print("Fahrenheit:", temps_f)
print("Hottest day:", days[np.argmax(temps_c)])
print("Difference from weekly mean:", temps_c - temps_c.mean())
```

### Exercise 1.6 — Aggregations along an axis

`scores` holds the marks of 5 students (rows) in 3 subjects (columns: Maths, Physics, Programming). Find:
1. the total marks of each student
2. the highest mark in each subject
3. the overall average of the whole class
4. which student (row number, counting from 1) has the highest total

```python
scores = np.array([[78, 65, 88],
                   [56, 72, 64],
                   [91, 85, 95],
                   [45, 50, 58],
                   [67, 74, 70]])

totals = scores.sum(axis=1)
print("Student totals:     ", totals)
print("Subject maximums:   ", scores.max(axis=0))
print("Class average:      ", round(scores.mean(), 2))
print("Top student (row):  ", np.argmax(totals) + 1)
```

### Exercise 1.7 — Broadcasting

Using the same `scores` matrix from 1.6:
1. Add grace marks of `[2, 0, 5]` to the three subjects in one operation.
2. Centre each subject by subtracting that subject's mean, so every column then has a mean of 0.
3. Scale each student's marks to a percentage of that student's own total. Each row should then add up to 100.

```python
grace = np.array([2, 0, 5])
print("With grace marks:\n", scores + grace, end="\n\n")

centred = scores - scores.mean(axis=0)
print("Centred:\n", centred)
print("Column means after centring:", centred.mean(axis=0), end="\n\n")

share = scores / scores.sum(axis=1, keepdims=True) * 100
print("Share of own total (%):\n", share)
print("Row sums:", share.sum(axis=1))
```

> **Note:** Broadcasting stretches the smaller array across the larger one when the trailing dimensions are equal or one of them is 1. `(5, 3) + (3,)` works because both end in 3. `(5, 3) / (5, 1)` works because 1 can stretch to 3.

### Exercise 1.8 — Conditional labels with np.where

Using `marks` from 1.4, create an array that holds `'Pass'` for marks ≥ 40 and `'Fail'` otherwise. Then make a second array of grades: `'A'` for ≥ 75, `'B'` for 60–74, `'C'` for 40–59 and `'F'` below 40.

```python
result = np.where(marks >= 40, "Pass", "Fail")
print(result)

conditions = [marks >= 75, marks >= 60, marks >= 40]
grades = np.select(conditions, ["A", "B", "C"], default="F")
print(grades)
```

### Exercise 1.9 — Random data and descriptive statistics

Set a random seed with `np.random.default_rng(7)`. Simulate 30 days of maximum temperature from a normal distribution with mean 32 °C and standard deviation 2.5 °C, rounded to one decimal. Report:
- mean, median, standard deviation, minimum and maximum
- on which day (1–30) the maximum occurred
- how many days were above 35 °C

```python
rng = np.random.default_rng(7)
temps = rng.normal(32, 2.5, 30).round(1)

print("Mean:   ", temps.mean().round(2))
print("Median: ", np.median(temps))
print("Std dev:", temps.std().round(2))
print("Min/Max:", temps.min(), "/", temps.max())
print("Hottest day:", np.argmax(temps) + 1)
print("Days above 35 °C:", (temps > 35).sum())
```

> **Note:** Using a fixed seed makes the 'random' numbers the same every time you run the cell, so your results can be reproduced.

### Exercise 1.10 — Why vectorise? (loop vs NumPy)

Create an array of one million random numbers. Compute the sum of their squares twice: once with a Python `for` loop and once with NumPy. Time both with `%timeit` (or `time.perf_counter`) and check that the answers match.

```python
import time

values = np.random.default_rng(0).random(1_000_000)

start = time.perf_counter()
total_loop = 0.0
for v in values:
    total_loop += v * v
loop_time = time.perf_counter() - start

start = time.perf_counter()
total_np = np.sum(values ** 2)
np_time = time.perf_counter() - start

print(f"Loop : {total_loop:.4f}  in {loop_time:.4f} s")
print(f"NumPy: {total_np:.4f}  in {np_time:.4f} s")
print("Same answer:", np.isclose(total_loop, total_np), "| NumPy is about", round(loop_time / np_time), "x faster")
```

## Key takeaways
- NumPy operations act on whole arrays at once (vectorisation). They are shorter and many times faster than loops.
- Boolean masks such as `a[a > 40]` are the standard way to filter.
- `axis=0` aggregates down the rows (one result per column). `axis=1` aggregates across the columns (one result per row).
- `info()` and `describe()` should be the first things you run on any new dataset. They show missing values and
  impossible values right away.
