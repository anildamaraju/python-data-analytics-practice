```python
# 1. Broadcasting
NumPy can operate on arrays of different shapes without loops. 
It "stretches" the smaller array when shapes are compatible. 
Compare shapes from the right: each pair of dimensions must be equal, or one of them must be 1.

# Problem Statement : How much each student's score differs from the average score in each subject.
```

```python
 # shape (3, 3)
marks = np.array([[70, 80, 90],
                  [60, 75, 85],
                  [80, 85, 95]])     
print(marks.shape)
```

```python
# Calculate Column Mean
#Axis 0 means go vertically (down each column).
# (70+60+80)/3 =210/3==70

col_mean = marks.mean(axis=0)             # shape (3,)  -> [70, 80, 90]
print(col_mean)
print(col_mean.shape)                     # It is a 1-D array containing one average for each subject
```

```python
#Subtract the Mean
# Broadcasting automatically expands the smaller array to match the larger array's shape.
#Subtracting the column mean from each row is called mean centering, a common preprocessing step in data analysis and machine learning.
# This tells us how much each student's score differs from the average score in each subject.
centered = marks - col_mean               # (3,3) - (3,) works
print(centered)
```
